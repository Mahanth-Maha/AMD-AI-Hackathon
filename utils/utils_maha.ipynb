```python

import os
import json

def add_topic_key(data, topic_value="SA"):
    for item in data:
        item["topic"] = topic_value
    return data
# Example usage after loading json_data:
# add_topic_key(json_data)

def create_json_train_test_from_jsons_in_a_dir(directory: str, json_prefix:str, train_out: str, test_out: str, split_ratio,topic_name ) -> None:
    """
    Create a clubbed JSON file from all JSON files in the specified directory.
    The output file will contain a JSON object with two keys: "train" and "test",
    each containing a list of question objects from the respective JSON files.
    """
    # combine all JSON files in the directory
    json_data = []
    for filename in os.listdir(directory):
        if filename.endswith(".json") and filename.startswith(json_prefix):
            file_path = os.path.join(directory, filename)
            with open(file_path, 'r', encoding='utf-8') as f:
                data = json.load(f)
                json_data.extend(data)
    # Add topic key to each item in the combined data
    json_data = add_topic_key(json_data, topic_name)
    
    # Split the data into train and test sets
    split_index = int(len(json_data) * split_ratio)
    train_data = json_data[:split_index]
    test_data = json_data[split_index:]
    # Write the output to the specified files 
    with open(train_out, 'w', encoding='utf-8') as train_file:
        json.dump(train_data, train_file, ensure_ascii=False, indent=4)
    with open(test_out, 'w', encoding='utf-8') as test_file:
        json.dump(test_data, test_file, ensure_ascii=False, indent=4)
        
create_json_train_test_from_jsons_in_a_dir(
    directory="/jupyter-tutorial/AAIPL_maha/gen_data_auto/BR",  
    json_prefix="br_dataset_",
    train_out="/jupyter-tutorial/AAIPL_maha/gen_data_auto/BR/br_train.json",
    test_out="/jupyter-tutorial/AAIPL_maha/gen_data_auto/BR/br_test.json",
    split_ratio=0.8,
    topic_name = "Puzzles involving generations and family tree logic"
)

create_json_train_test_from_jsons_in_a_dir(
    directory="/jupyter-tutorial/AAIPL_maha/gen_data_auto/TL",  
    json_prefix="people",
    train_out="/jupyter-tutorial/AAIPL_maha/gen_data_auto/TL/tl_train.json",
    test_out="/jupyter-tutorial/AAIPL_maha/gen_data_auto/TL/tl_test.json",
    split_ratio=0.8,
    topic_name = "Truth-teller and Liar Problems"
)    

create_json_train_test_from_jsons_in_a_dir(
    directory="/jupyter-tutorial/AAIPL_maha/gen_data_auto/SA",  
    json_prefix="sa_dataset",
    train_out="/jupyter-tutorial/AAIPL_maha/gen_data_auto/SA/sa_train.json",
    test_out="/jupyter-tutorial/AAIPL_maha/gen_data_auto/SA/sa_test.json",
    split_ratio=0.8,
    topic_name = "Seating Arrangements (Linear, Circular)"
)    
```
