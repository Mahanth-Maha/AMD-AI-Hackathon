```python
def extract_text_from_pdf(filepath):
    text = []
    with pdfplumber.open(filepath) as pdf:
        for page in pdf.pages:
            visible_area = page.cropbox if page.cropbox else page.mediabox
            cropped_page = page.within_bbox(visible_area)
            text.append(cropped_page.extract_text())
    return "\n".join(text)

def extract_text_from_csv(filepath):
    """Extracts text from a CSV file by joining rows as tab-separated lines, with each value preceded by its column name."""
    text_lines = []
    try:
        with open(filepath, 'r', encoding='utf-8', errors='replace') as csvfile:
            reader = csv.reader(csvfile)
            headers = next(reader, None)
            if headers is None:
                return ''
            for row in reader:
                # Pair each value with its header: Header:Value
                row_with_headers = [f"{header}:  {value}" for header, value in zip(headers, row)]
                text_lines.append('\t'.join(row_with_headers))
    except Exception as e:
        print(f"Error reading CSV {filepath}: {str(e)}")
    return '\n'.join(text_lines)
```

```python
def extract_text_from_txt(filepath):
    text = []
    with open(filepath) as f:
        for line in f:
            if line.strip():
                text.append(line.strip())
                
    return "\n".join(text)
```
